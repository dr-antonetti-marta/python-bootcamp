Python's basic types are integers (`int`), decimals (`float`), text (`str`) and `True`/`False` (`bool`). Later we'll meet *categorical* variables (what R calls factors).

Now collections. A **list** holds several values in order; a **NumPy array** is a list built for maths: it does arithmetic on every element at once (R users: this is the closest thing to an R vector).

````python
import numpy as np

a_list = [1, 2, 3]            # a plain Python list
a_sequence = np.array([1, 2, 3])   # a NumPy array

print(type(a_list), type(a_sequence))
print(a_sequence.sum())
print(2 * a_sequence)         # element-by-element
print(2 * a_list)             # NOT the same: a list "times 2" is the list repeated twice!
print(len(a_sequence))        # length

abc = np.array(["a", "b", "c"])
# An array holds one type of element. Mixing numbers and text makes everything text:
print(abc.dtype)
print(np.array([1, "b", 3]))
````

> **R → Python:** Python counts from **0**, not 1. The first element is `x[0]`, and a slice like `x[0:2]` stops *before* index 2 (so it returns two elements). This is the single most common mistake for people coming from R, so keep it in mind in everything below.

### Sequences & arrays *(syntax demo)*

````python
another_sequence = np.arange(0.5, 1.5 + 0.1, 0.5)   # start, stop (excluded!), step -- so we nudge the stop past 1.5
print(another_sequence)
print(np.linspace(0.5, 1.5, 3))                      # alternative: start, stop (included), how many values

third_sequence = np.arange(1, 4)                     # 1, 2, 3 (the stop value 4 is excluded)
print(third_sequence)

rep1 = np.repeat(1, 5)                               # 1 1 1 1 1
print(rep1)

rep2 = np.tile([1, 2, 3], 3)                         # whole pattern repeated: 1 2 3 1 2 3 1 2 3
print(rep2)

rep3 = np.repeat([1, 2, 3], 3)                       # each value repeated: 1 1 1 2 2 2 3 3 3
print(rep3)
````

````python
# Let's have an insight on how Python thinks
matrix = np.column_stack([a_sequence, abc])   # columns side by side (R: cbind)
print(matrix)
print(matrix[1, 1])        # row index 1, column index 1 -> the SECOND row, SECOND column
# How do I get "3"? and how do I get "a"?  (remember: counting starts at 0)

matrix2 = np.vstack([a_sequence, abc])        # stack as rows (R: rbind)
print(matrix2)
print(matrix2[1, 2])
# How do I get "a" now?
# If I mix up rows and columns and ask for matrix2[3, 1], I get an error:
# IndexError: index 3 is out of bounds for axis 0 with size 2
````

### Working directories and finding files *(syntax demo)*

Almost every beginner problem with data comes down to one question: **where does Python think it is?** Relative paths such as `"data/Seatbelts.csv"` are interpreted from the notebook's *current working directory*, and that is not always the folder you expect (it depends on how you launched Jupyter or your editor).

````python
from pathlib import Path

print(Path.cwd())           # where Python is looking right now
print(Path.cwd().parent)    # the folder above it (".." in a path means "go up one folder")
````

Rather than hoping the working directory is right, we'll make our code **find its data**. The helper below looks for a `data` folder in the current folder, then in the folders above it. (You can use it as it is; you don't need to understand every line yet.) It is repeated at the top of every session.

````python
from pathlib import Path
import calendar

import pandas as pd

MONTHS = list(calendar.month_name)[1:]   # ["January", "February", ..., "December"]


def find_data_file(name):
    """Look for data/<name> in this folder, in the folders above it, and one level below."""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        candidate = folder / "data" / name
        if candidate.exists():
            return candidate
    for candidate in sorted(here.glob(f"*/data/{name}")):
        return candidate
    raise FileNotFoundError(
        f"Could not find data/{name}. Python is currently looking from: {here}\n"
        "Fix: put the 'data' folder next to this notebook (or in a folder above it)."
    )


print(find_data_file("Seatbelts.csv"))
````

### Importing data

`Seatbelts` is real historical data (Great Britain road-safety statistics, 1969-1984). In R it ships built in; in Python we read it from a plain CSV file in the `data` folder. It comes from R's `datasets` package (Harvey & Durbin 1986; Harvey 1989).

````python
seatbelts = pd.read_csv(find_data_file("Seatbelts.csv"))

# Some ways of inspecting the dataset
print(seatbelts.shape)        # (rows, columns)
seatbelts.head()              # the first 5 rows (change the number: .head(10))
````

Notice there is no column telling us *when* each row happened: R stores this dataset as a time series object that carries its own dates, and the CSV loses them. We know the data run monthly from January 1969 to December 1984 (192 months), so we can rebuild the dates ourselves, and add the month's name:

````python
seatbelts_monthly = seatbelts.copy()
seatbelts_monthly["date"] = pd.date_range("1969-01-01", periods=len(seatbelts_monthly), freq="MS")   # MS = month start
seatbelts_monthly["month"] = seatbelts_monthly["date"].dt.month_name()

seatbelts_monthly.head()
````

````python
# Give "month" its natural calendar order (otherwise Python would sort the names alphabetically).
# A column with a fixed set of ordered values is a *categorical* variable -- R calls this a factor.
seatbelts_monthly["month"] = pd.Categorical(seatbelts_monthly["month"], categories=MONTHS, ordered=True)

seatbelts_monthly.dtypes
````

````python
# law: 0 before the seatbelt law, 1 after. The law was introduced on 31 January 1983,
# so the first month flagged 1 is February 1983.
print(seatbelts_monthly["law"].value_counts().sort_index())
print(seatbelts_monthly.loc[seatbelts_monthly["law"] == 1, "date"].min())
````

> **R → Python:** `.copy()` matters. In R, `b <- a` gives you an independent copy. In Python, `b = a` just gives the *same object* a second name, so changing `b` also changes `a`. Whenever you want a separate version of a data frame, say so explicitly with `.copy()`.

What if my data set had a different extension, say it was created in Stata? pandas can read and write many formats. Here we write our own data out as a `.dta` file and read it straight back in, just to see the mechanics:

````python
data_dir = find_data_file("Seatbelts.csv").parent       # the folder our data lives in

seatbelts_monthly.to_stata(data_dir / "seatbelts_monthly_demo.dta", write_index=False)   # write our own data out as .dta...
seatbelts_from_stata = pd.read_stata(data_dir / "seatbelts_monthly_demo.dta")            # ...and read it straight back in
seatbelts_from_stata.info()
````

### Subsetting

````python
## (1) Square brackets and boolean conditions

# Select the first 10 observations (positions 0 to 9)
seatbelts_10 = seatbelts_monthly.iloc[:10]
seatbelts_10
````

````python
# Select columns (note the double brackets: a list of column names)
seatbelts_cols = seatbelts_monthly[["date", "DriversKilled", "law"]]
seatbelts_cols.head()
````

````python
# Select rows by a condition: build a True/False series, then use it as a filter
seatbelts_postlaw = seatbelts_monthly[seatbelts_monthly["law"] == 1]
print(seatbelts_postlaw.shape)   # only 23 months -- the law took effect on 31 Jan 1983, and our data ends Dec 1984
````

````python
## (2) .query() and method chaining
# .query() lets you write the condition as text; chaining methods one after another reads top to bottom.

# Select columns
seatbelts_select = seatbelts_monthly[["date", "DriversKilled", "kms", "law"]]

# Select rows
seatbelts_1980s = seatbelts_monthly.query("date >= '1980-01-01'")

# Chain operations -- and here's an important design decision, not just a coding one:
# comparing ALL 169 pre-law months against the 23 post-law months would mostly be
# comparing two very different eras (cars, roads and driving all got safer over 15
# years, law or no law) -- not the law's effect. Narrowing the pre-law window keeps
# the comparison closer to apples-to-apples. We'll use this same window for the rest
# of the bootcamp whenever we compare before/after the law.
seatbelts_window = (
    seatbelts_monthly
    .query("date >= '1980-01-01'")
    .loc[:, ["date", "DriversKilled", "kms", "PetrolPrice", "law"]]
)
print(seatbelts_window["law"].value_counts().sort_index())
# can you produce a narrower window, say 2 years either side of the law change instead of
# our 3-years-before choice? how many observations in each group do you get, and does the
# comparison feel any more or less trustworthy to you?
````

### Rename variables *(syntax demo)*

````python
# before renaming variables: a column is created (or overwritten) just by assigning to its name
seatbelts_monthly_original = seatbelts_monthly.copy()   # keep an untouched copy so the alternatives below start from the same data

seatbelts_monthly["deaths"] = seatbelts_monthly["DriversKilled"]       # a NEW column; the old one is still there
seatbelts_monthly["distance_km"] = seatbelts_monthly["kms"]
print(list(seatbelts_monthly.columns))    # we can check what really happened
````

````python
# .rename() really renames: old name -> new name (it returns a new data frame; the original is unchanged)
seatbelts_renamed = seatbelts_monthly_original.rename(columns={"DriversKilled": "deaths", "kms": "distance_km"})
seatbelts_renamed.head()   # see that PetrolPrice did not change
````

````python
# Or rename in place on the data frame itself (inplace=True changes the original object):
seatbelts_base = seatbelts_monthly_original.copy()
seatbelts_base.rename(columns={"DriversKilled": "deaths", "kms": "distance_km"}, inplace=True)
print(list(seatbelts_base.columns))

# Another common pattern: assign a whole new list of names (must be the right length and order!)
# seatbelts_base.columns = ["a", "b", ...]
````

### Omit NAs *(syntax demo)*

````python
# Seatbelts itself has no missing data, so for teaching purposes we'll pretend a handful of
# months' PetrolPrice reading went missing
rng = np.random.default_rng(1)                        # a random number generator, seeded so everyone gets the same result
missing_rows = rng.choice(len(seatbelts_monthly), size=5, replace=False)
seatbelts_monthly.loc[seatbelts_monthly.index[missing_rows], "PetrolPrice"] = np.nan   # NaN = "not a number" = Python's missing value

print(seatbelts_monthly["PetrolPrice"].isna().sum(), "missing values")
seatbelts_clean = seatbelts_monthly.dropna()
# how many observations did we lose?
print(seatbelts_monthly.shape[0] - seatbelts_clean.shape[0])
````
